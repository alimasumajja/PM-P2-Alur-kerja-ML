```python
#BAGIAN C langkah latihan
import pandas as pd
data = {'luas':[36,70,45,90,54], 'kamar':[2,3,2,4,3],
        'lokasi':['Pinggir','Tengah','Pinggir','Tengah','Pinggir'],
        'harga':[250,600,320,800,400]}
df = pd.DataFrame(data)
X = df[['luas','kamar','lokasi']]
y = df['harga']
print(df['harga'].tolist())
df['kategori'] = ['Murah','Mahal','Murah','Mahal','Murah']

from sklearn.model_selection import train_test_split
X_train,X_test,y_train,y_test = train_test_split(
    X, y, test_size=0.4, random_state=42)
print('Latih:', len(X_train), ' Uji:', len(X_test))
```
